## **1. Subespacios Vectoriales en $\mathbb{R}^n$**

Un subconjunto $W$ de $\mathbb{R}^n$ se denomina **subespacio vectorial** si la estructura es cerrada bajo las operaciones usuales de adición y multiplicación por un escalar. Es decir, satisface las siguientes dos propiedades para todo $\mathbf{u}, \mathbf{v} \in W$ y cualquier $c \in \mathbb{R}$:

1. **Clausura bajo la suma:** $\mathbf{u} + \mathbf{v} \in W$
2. **Clausura bajo el producto escalar:** $c\mathbf{u} \in W$

> 💡 **Nota:** Todo subespacio vectorial válido debe contener obligatoriamente al vector cero ($\mathbf{0}$), ya que si hacemos $c = 0$ en la segunda propiedad, $0\mathbf{u} = \mathbf{0} \in W$.

## **2. Combinación Lineal y Conjunto Generador ($\text{gen } S$)**

Si tenemos una lista de $k$ vectores $\mathbf{a}_1, \mathbf{a}_2, \dots, \mathbf{a}_k$ en $\mathbb{R}^m$ y un conjunto de escalares $x_1, x_2, \dots, x_k$, el vector resultante:
$$\mathbf{b} = x_1\mathbf{a}_1 + x_2\mathbf{a}_2 + \dots + x_k\mathbf{a}_k$$
se conoce como una **combinación lineal**. Esta operación puede compactarse como un producto matriz-vector $A\mathbf{x} = \mathbf{b}$, donde las columnas de $A$ son nuestros vectores.

El conjunto de *todas* las combinaciones lineales posibles de los vectores en $S = \{\mathbf{a}_1, \dots, \mathbf{a}_k\}$ forma un subespacio de $\mathbb{R}^m$ y se denomina **conjunto generado por $S$** o $\text{gen } S$.

## **3. Independencia Lineal**

Un conjunto de vectores $\mathbf{a}_1, \mathbf{a}_2, \dots, \mathbf{a}_k$ es **linealmente independiente (L.I.)** si y solo si la única forma de obtener el vector nulo mediante su combinación lineal es que todos los coeficientes sean cero:
$$x_1\mathbf{a}_1 + x_2\mathbf{a}_2 + \dots + x_k\mathbf{a}_k = \mathbf{0} \implies x_1 = x_2 = \dots = x_k = 0$$


*   **Propiedad fundamental:** Cualquier conjunto de vectores que contenga al vector cero ($\mathbf{0}$) es automáticamente dependiente.

Para entender esto, analizaremos detalladamente dos ejemplos concretos con sus respectivas matrices:

### **Ejemplo 1**
Analicemos los siguientes tres vectores en $\mathbb{R}^3$:
$$\mathbf{u} = \begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix}, \quad \mathbf{v} = \begin{bmatrix} 1 \\ 0 \\ -1 \end{bmatrix}, \quad \mathbf{w} = \begin{bmatrix} 2 \\ 1 \\ 0 \end{bmatrix}$$
*Nota analítica:* Observa que el tercer vector es la suma exacta de los dos primeros ($\mathbf{w} = \mathbf{u} + \mathbf{v}$). Esto significa que existe información redundante. Si colocamos estos vectores como las columnas de una matriz $A$, el sistema homogéneo  $A\mathbf{x} = \mathbf{0}$ tendrá soluciones distintas de cero.

$$A= \begin{bmatrix} 1 & 1 & 2 \\ 1 & 0 & 1 \\ 1 & -1 & 0 \end{bmatrix}$$

### **Ejemplo 2**
Ahora, analicemos este otro conjunto de vectores en $\mathbb{R}^3$:
$$\mathbf{u} = \begin{bmatrix} 3 \\ 0 \\ 0 \end{bmatrix}, \quad \mathbf{v} = \begin{bmatrix} 0 \\ 1 \\ -1 \end{bmatrix}, \quad \mathbf{w} = \begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix}$$
*Nota analítica:* Ninguno de estos vectores puede ser construido a partir de combinaciones de los otros. Si los agrupamos en la matriz $A_{ind}$, el único vector de coeficientes que satisface $A\mathbf{x} = \mathbf{0}$ es el vector nulo.

$$A = \begin{bmatrix} 3 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & -1 & 1 \end{bmatrix}$$

```python
# Verificación de la Independencia Lineal de los Ejemplos 1 y 2

import numpy as np

# Definimos las matrices exactamente como se describieron en el texto superior
A_dep = np.array([
    [1,  1, 2],
    [1,  0, 1],
    [1, -1, 0]
])

A_ind = np.array([
    [3, 0, 0],
    [0, 1, 1],
    [0, -1, 1]
])

# Calculamos el rango (número de columnas linealmente independientes reales)
rango_dep = np.linalg.matrix_rank(A_dep)
rango_ind = np.linalg.matrix_rank(A_ind)

print("--- ANÁLISIS DEL EJEMPLO A (DEPENDIENTE) ---")
print(f"Matriz A_dep:\n{A_dep}")
print(f"Número de vectores (columnas): {A_dep.shape[1]}")
print(f"Vectores linealmente independientes detectados (Rango): {rango_dep}")
print(f"¿El conjunto es Linealmente Independiente?: {rango_dep == A_dep.shape[1]}")

print("\n" + "="*50 + "\n")

print("--- ANÁLISIS DEL EJEMPLO B (INDEPENDIENTE) ---")
print(f"Matriz A_ind:\n{A_ind}")
print(f"Número de vectores (columnas): {A_ind.shape[1]}")
print(f"Vectores linealmente independientes detectados (Rango): {rango_ind}")
print(f"¿El conjunto es Linealmente Independiente?: {rango_ind == A_ind.shape[1]}")
```

Antes de definir formalmente lo que es una base, analicemos dos restricciones dimensionales cruciales:
1. **Teorema del límite superior:** Si un conjunto de vectores en $\mathbb{R}^n$ es linealmente independiente, puede tener a lo sumo $n$ elementos ($k \leq n$). Por lo tanto, cualquier conjunto de 3 vectores en $\mathbb{R}^2$ es obligatoriamente dependiente.
2. **Teorema del límite inferior:** Si un conjunto de vectores genera todo el espacio $\mathbb{R}^n$, debe tener al menos $n$ elementos ($k \geq n$). Por ende, 2 vectores en $\mathbb{R}^3$ jamás podrán generar todo el espacio.

### **Bases en $\mathbb{R}^n$**
Una **base** de $\mathbb{R}^n$ es un conjunto de exactamente $n$ vectores que son linealmente independientes y que, en consecuencia, tienen la capacidad de generar todo $\mathbb{R}^n$.

*   **Expansión Única:** Si un conjunto $B = \{\mathbf{a}_1, \dots, \mathbf{a}_n\}$ es una base, cualquier vector $\mathbf{b} \in \mathbb{R}^n$ se puede expresar de forma **única** como:
    $$x_1\mathbf{a}_1 + x_2\mathbf{a}_2 + \dots + x_n\mathbf{a}_n = \mathbf{b}$$
    A los escalares $x_i$ se les llama los coeficientes de la expansión de $\mathbf{b}$ en la base $B$.





#### **Ejemplo**
Queremos expandir el vector $\mathbf{b} = \begin{bmatrix} 5 \\ 3 \end{bmatrix}$ en la base de $\mathbb{R}^2$ dada por:
$$B = \left\{ \mathbf{a}_1 = \begin{bmatrix} 2 \\ 1 \end{bmatrix}, \, \mathbf{a}_2 = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \right\}$$

Esto se traduce en resolver el sistema:
$$\begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix} \begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = \begin{bmatrix} 5 \\ 3 \end{bmatrix}$$

```python
# Resolución del Ejemplo de Expansión en una Base no Canónica
import numpy as np

# Definimos la matriz base A (con los vectores a1 y a2 como columnas)
A_base = np.array([
    [2, 1],
    [1, 2]
])

# Definimos el vector b que queremos expandir
b = np.array([5, 3])

# Resolvemos para encontrar los coeficientes únicos x1 y x2
coeficientes = np.linalg.solve(A_base, b)

print("--- EXPANSIÓN EN UNA BASE ---")
print(f"Coeficiente x1 (para a1): {coeficientes[0]:.2f}")
print(f"Coeficiente x2 (para a2): {coeficientes[1]:.2f}")
print(f"Verificación (x1*a1 + x2*a2): {coeficientes[0]}*A_base[:,0] + {coeficientes[1]}*A_base[:,1] = {np.dot(A_base, coeficientes)}")
```

## **5. El Rango de una Matriz y la Factorización $A = CR$**

El **rango** de una matriz $A$ (notado como $r$) es el número de columnas (o filas) linealmente independientes que posee. Este número define con precisión la dimensión del espacio generado por sus columnas.

Podemos descomponer cualquier matriz $A$ en el producto de una matriz $C$ (columnas independientes extraídas de $A$) y una matriz $R$ (la forma escalonada reducida purgada de filas nulas):
$$A = CR$$


Para entender cómo varía el rango y qué significa geométricamente, analicemos este ejemplo en $\mathbb{R}^3$:

### **Ejemplo**
$$A_1 = \begin{bmatrix} 1 & 3 & 8 \\ 1 & 2 & 6 \\ 0 & 1 & 2 \end{bmatrix}$$
 Aunque la matriz tiene $3$ columnas, la tercera es combinación lineal de las dos primeras ($\text{col}_3 = 2\cdot\text{col}_1 + 2\cdot\text{col}_2$). Al haber solo $2$ columnas linealmente independientes, el espacio generado es un plano bidimensional que pasa por el origen en $\mathbb{R}^3$.

A continuación, realizaremos esta verificación computacional.

```python
import numpy as np
from sympy import Matrix

# Definimos la matriz del ejemplo
A_sympy = Matrix([[1, 3, 8],
                  [1, 2, 6],
                  [0, 1, 2]])

# 1. Obtener RREF y columnas pivote
Arref, pivotes = A_sympy.rref()

# 2. Construir la matriz C con las columnas originales independientes
C_sympy = A_sympy[:, pivotes]

# 3. Construir la matriz R (eliminando las filas llenas de ceros de la RREF)
# Convertimos primero a NumPy para manipulación de filas de forma simple
R_full = np.array(Arref, dtype=float)
R_np = R_full[~np.all(np.isclose(R_full, 0), axis=1)]

# Convertimos C a numpy para operar libremente
C_np = np.array(C_sympy, dtype=float)

print(f"Matriz original A:\n{np.array(A_sympy, dtype=float)}\n")
print(f"Matriz C (Columnas L.I.):\n{C_np}\n")
print(f"Matriz R (RREF simplificada):\n{R_np}\n")

# 4. Reconstrucción por producto matricial: A = C * R
A_reconstruida = np.dot(C_np, R_np)
print(f"Producto CR (Debe ser igual a A):\n{A_reconstruida}")
```

## **6. Los Cuatro Subespacios Fundamentales**

Cada matriz $A \in \mathbb{R}^{m \times n}$ tiene asociados cuatro subespacios vectoriales intrínsecos:

1. **Espacio Columna ($C(A) \subset \mathbb{R}^m$):** El espacio generado por las columnas de $A$. Su dimensión es el rango $r$.
2. **Espacio Fila ($C(A^T) \subset \mathbb{R}^n$):** El espacio generado por las filas de $A$. Su dimensión es también el rango $r$.
3. **Espacio Nulo ($N(A) \subset \mathbb{R}^n$):** El conjunto de soluciones al sistema homogéneo $A\mathbf{w} = \mathbf{0}$. Su dimensión es $n - r$.
4. **Espacio Nulo Izquierdo ($N(A^T) \subset \mathbb{R}^m$):** El conjunto de soluciones a $A^T\mathbf{y} = \mathbf{0}$. Su dimensión es $m - r$.
