## **1. Introducción al Ajuste de Datos**

Introducimos una de las aplicaciones más importantes del método de mínimos cuadrados: el **ajuste de datos** (Data Fitting). El objetivo principal es construir un modelo matemático aproximado que describa la relación subyacente entre variables a partir de un conjunto de datos observados.

### **El Problema General**
Supongamos que tenemos un vector de atributos $\mathbf{x} \in \mathbb{R}^n$ y un resultado escalar $y \in \mathbb{R}$, y asumimos que están relacionados mediante una función desconocida $f$:

$$y \approx f(\mathbf{x})$$

*   **Vector de atributos ($\mathbf{x}$):** También llamado vector de variables independientes, características o predictores.
*   **Variable de respuesta ($y$):** Representa el resultado observado de interés. Cuando los atributos corresponden a valores pasados de una serie temporal e $y$ es el valor inmediato siguiente, el problema se conoce como **regresión**.

---

## **2. Datos y Modelo**

### **Los Datos**
Aunque no conocemos la forma exacta de $f$, disponemos de un conjunto de $N$ observaciones empíricas:

$$\mathbf{x}^{(1)}, \dots, \mathbf{x}^{(N)}, \qquad y^{(1)}, \dots, y^{(N)}$$

Donde el superíndice $(i)$ denota la $i$-ésima muestra o experimento, $\mathbf{x}^{(i)}$ es un $n$-vector de características y el escalar $y^{(i)}$ es su resultado real medido.

### **El Modelo**
Proponemos una función de predicción aproximada $\hat{f}$ para estimar el valor del resultado:

$$\hat{y} = \hat{f}(\mathbf{x})$$

El símbolo de sombrero ($\,\hat{}\,$) indica que estamos ante una estimación o aproximación matemática del fenómeno real.

## **3. Modelos Lineales en los Parámetros**

Nos enfocaremos en modelos que adoptan una combinación lineal de ciertas funciones base:

$$\hat{f}(\mathbf{x}) = \theta_1 f_1(\mathbf{x}) + \theta_2 f_2(\mathbf{x}) + \dots + \theta_p f_p(\mathbf{x})$$

Donde:
*   $f_j: \mathbb{R}^n \to \mathbb{R}$ son **funciones base** preseleccionadas por el analista según la naturaleza del problema.
*   $\theta_j$ son los **parámetros o coeficientes** del modelo que debemos determinar de forma óptima.

Este modelo se denomina *lineal en los parámetros* porque, para cualquier vector de entrada $\mathbf{x}$, la predicción $\hat{y}$ es una función lineal del vector de coeficientes $\mathbf{\theta} = [\theta_1, \dots, \theta_p]^T$.

---

## **4. Formulación Matricial del Problema**

Para cada una de las $N$ observaciones en nuestro conjunto de datos, podemos expresar la predicción del modelo como:

$$\hat{y}^{(i)} = A_{i1}\theta_1 + A_{i2}\theta_2 + \dots + A_{ip}\theta_p, \qquad i = 1, \dots, N$$

Definiendo la **matriz de diseño** $A \in \mathbb{R}^{N \times p}$ a partir de las funciones base evaluadas en cada dato:

$$A_{ij} = f_j(\mathbf{x}^{(i)}), \qquad i = 1, \dots, N, \quad j = 1, \dots, p$$

El sistema completo de predicciones adopta la forma compacta:

$$\mathbf{\hat{y}}^d = A\mathbf{\theta}$$

El vector residual de errores de predicción es $\mathbf{r}^d = \mathbf{y}^d - A\mathbf{\theta}$. Nuestro objetivo es encontrar el vector de parámetros $\mathbf{\theta}$ que minimice la **Suma de Cuadrados de los Residuos (RSS)**, lo que nos conduce al problema clásico de mínimos cuadrados:

$$\min_{\mathbf{\theta} \in \mathbb{R}^p} \|A\mathbf{\theta} - \mathbf{y}^d\|^2$$

Si las columnas de $A$ son linealmente independientes, la solución única se halla resolviendo el **Sistema de Ecuaciones Normales**:

$$A^TA\mathbf{\hat{\theta}} = A^T\mathbf{y}^d$$

---

### **Métricas de Rendimiento del Ajuste**
*   **Mínimo Error Cuadrático Medio (MMSE):** Denota el promedio del error al cuadrado en las muestras: $\text{MMSE} = \frac{1}{N}\|A\mathbf{\hat{\theta}} - \mathbf{y}^d\|^2$.
*   **Error de Ajuste RMS:** Es la raíz cuadrada del MMSE, indicando la desviación típica de las predicciones en las mismas unidades que la variable de respuesta.

## **5. Ajuste Lineal (Afín) por Mínimos Cuadrados**

En el caso más simple de una sola variable independiente ($n=1$), seleccionamos como funciones base $f_1(x) = 1$ (término de intercepto) y $f_2(x) = x$ (término de pendiente). El modelo toma la forma de una línea recta:

$$\hat{f}(x) = \theta_1 + \theta_2 x$$

La matriz de diseño $A$, el vector de parámetros $\mathbf{\theta}$ y el vector de observaciones reales $\mathbf{y}^d$ se estructuran de la siguiente manera:

$$A = \begin{bmatrix} 1 & x^{(1)} \\ 1 & x^{(2)} \\ \vdots & \vdots \\ 1 & x^{(N)} \end{bmatrix}, \quad \mathbf{\theta} = \begin{bmatrix} \theta_1 \\ \theta_2 \end{bmatrix}, \quad \mathbf{y}^d = \begin{bmatrix} y^{(1)} \\ y^{(2)} \\ \vdots \\ y^{(N)} \end{bmatrix}$$

Si los puntos $x^{(i)}$ no son todos idénticos, las columnas de $A$ son linealmente independientes y los parámetros óptimos se calculan mediante:

$$\mathbf{\hat{\theta}} = (A^TA)^{-1}A^T\mathbf{y}^d$$

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
URL1 = 'https://raw.githubusercontent.com/NumericalMethodsSullivan'
URL2 = '/NumericalMethodsSullivan.github.io/master/data/'
URL = URL1+URL2
data = np.array( pd.read_csv(URL+'Exercise4_51.csv') )
# Exercise4_51.csv

x = data[:,0]
y = data[:,1]
plt.plot(x,y,'bo',markersize=5)
```

```python
import numpy as np
from scipy import linalg
from sympy import *
import matplotlib.pyplot as plt

A =np.vstack((np.ones(len(x)), x)).T

b = np.array(y)
print("A =",A,"\n")
print("b =",b,"\n")

# Solución por mínimos cuadrados de sistemas lineales m x n
theta = np.linalg.lstsq(A, b.reshape(-1, 1), rcond=None)[0]

yhat = A.dot(theta)
print("x =",theta,"\n")

print("Ax =",yhat,"\n")

# plot data and predictions
plt.scatter(x, y)
plt.plot(x, yhat, color='red')
```

## **6. Ajuste Polinomial por Mínimos Cuadrados**

El ajuste polinomial es una generalización directa del caso lineal afín. En lugar de limitarnos a una línea recta, seleccionamos como funciones base potencias sucesivas de la variable independiente $x$:

$$f_k(x) = x^{k-1}, \qquad k = 1, \dots, p$$

De esta manera, el modelo toma la forma de un polinomio de grado menor o igual a $k-1$:

$$\hat{f}(x) = \theta_1 + \theta_2 x + \theta_3 x^2 + \dots + \theta_k x^{k-1}$$

### **Estructura Matricial (Matriz de Vandermonde)**
Para un conjunto de $N$ observaciones, la matriz de diseño $A$, el vector de parámetros $\mathbf{\theta}$ y el vector de respuestas reales $\mathbf{y}^d$ se definen como:

$$A = \begin{bmatrix}
1 & x^{(1)} & (x^{(1)})^2 & \dots & (x^{(1)})^{k-1} \\
1 & x^{(2)} & (x^{(2)})^2 & \dots & (x^{(2)})^{k-1} \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
1 & x^{(N)} & (x^{(N)})^2 & \dots & (x^{(N)})^{k-1}
\end{bmatrix}, \quad \mathbf{\theta} = \begin{bmatrix}
\theta_1 \\
\theta_2 \\
\vdots \\
\theta_k
\end{bmatrix}, \quad \mathbf{y}^d = \begin{bmatrix}
y^{(1)} \\
y^{(2)} \\
\vdots \\
y^{(N)}
\end{bmatrix}$$

## **7. Ejemplo de Ajuste Cuadrático ($k=3$)**

A continuación, ajustaremos un modelo polinomial de segundo grado (parábola) a un nuevo conjunto de datos utilizando la función de mínimos cuadrados de NumPy.

```python
import numpy as np
import pandas as pd
URL1 = 'https://raw.githubusercontent.com/NumericalMethodsSullivan'
URL2 = '/NumericalMethodsSullivan.github.io/master/data/'
URL = URL1+URL2
data = np.array( pd.read_csv(URL+'Exercise4_52.csv') )
# Exercise4_52.csv

x = data[:,0]
x2 = np.square(x)
y = data[:,1]
plt.plot(x,y,'bo',markersize=5)
```

```python
A =np.vstack((np.ones(len(x)), x, x2)).T

b = np.array(y)
print("A =",A,"\n")
print("b =",b,"\n")

# Solución por mínimos cuadrados de sistemas lineales m x n
theta = np.linalg.lstsq(A, b.reshape(-1, 1), rcond=None)[0]

yhat = A.dot(theta)
print("x =",theta,"\n")

print("Ax =",yhat,"\n")

# plot data and predictions
plt.scatter(x, y)
plt.plot(x, yhat, color='red')
```

```python
# =====================================================================
# 8. Plantilla de Trabajo para Datasets de Prueba
# =====================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

URL_BASE = 'https://raw.githubusercontent.com/NumericalMethodsSullivan/NumericalMethodsSullivan.github.io/master/data/'

try:
    datasetA = np.array(pd.read_csv(URL_BASE + 'Exercise3_datafit2.csv'))
    datasetB = np.array(pd.read_csv(URL_BASE + 'Exercise3_datafit3.csv'))
    datasetC = np.array(pd.read_csv(URL_BASE + 'Exercise3_datafit4.csv'))
except Exception as e:
    print("Error cargando datasets externos. Usando conjuntos simulados para la demostración...")
    t = np.linspace(0, 6, 60)
    datasetA = np.column_stack((t, 3.0 + 2.5 * np.exp(t/2) + np.random.normal(0, 1.5, size=60)))
    datasetB = np.column_stack((t, 1.0 + 4.0 * np.sin(t) + 2.0 * np.cos(t) + np.random.normal(0, 0.5, size=60)))
    datasetC = np.column_stack((t, 5.0 - 1.2 * t + 0.8 * (t**2) + np.random.normal(0, 1, size=60)))

# ---------------------------------------------------------------------
# PASO 1: Visualización inicial de los tres datasets
# ---------------------------------------------------------------------
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

axs[0].scatter(datasetA[:, 0], datasetA[:, 1], color='purple', alpha=0.7, edgecolors='k')
axs[0].set_title('Dataset A - Analizar Tendencia')
axs[0].grid(True)

axs[1].scatter(datasetB[:, 0], datasetB[:, 1], color='green', alpha=0.7, edgecolors='k')
axs[1].set_title('Dataset B - Analizar Tendencia')
axs[1].grid(True)

axs[2].scatter(datasetC[:, 0], datasetC[:, 1], color='orange', alpha=0.7, edgecolors='k')
axs[2].set_title('Dataset C - Analizar Tendencia')
axs[2].grid(True)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------------------
# PASO 2: Ejemplo de Solución para el Dataset A
# (Modifica las columnas de A según tu conjetura matemática)
# ---------------------------------------------------------------------
x_A = datasetA[:, 0]
y_A = datasetA[:, 1]

# [CONJETURA EJEMPLO]: Intentemos un ajuste cuadrático básico como punto de partida
col_ones = np.ones(len(x_A))
col_x = x_A
col_x2 = np.square(x_A)

# Construir matriz A ensamblando las funciones base elegidas
A_A = np.vstack((col_ones, col_x, col_x2)).T
b_A = y_A.reshape(-1, 1)

# Resolver el sistema
theta_A = np.linalg.lstsq(A_A, b_A, rcond=None)[0]
yhat_A = A_A.dot(theta_A)

# Graficar resultados del Dataset A
plt.figure(figsize=(8, 5))
plt.scatter(x_A, y_A, color='purple', label='Datos Reales (A)')
plt.plot(x_A, yhat_A, color='red', linewidth=2, label='Curva de Ajuste Propuesta')
plt.title('Validación del Modelo Elegido para Dataset A')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()
```
