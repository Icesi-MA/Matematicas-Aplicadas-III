## 2. El problema del volumen en el espacio tridimensional

En los cursos previos de cálculo, los esfuerzos se concentraron principalmente en dos problemas fundamentales: la derivación (orientada a la optimización de variables) y la integración definida en una variable. Esta última respondía a la necesidad de encontrar la cantidad total de una magnitud acumulada en una región del plano, lo que geométricamente interpretábamos como el área bajo una curva.

Sin embargo, los fenómenos de la ingeniería suelen ser multifactoriales. Cuando trabajamos con funciones de varias variables, el interés matemático se traslada del área al volumen, o a la acumulación de magnitudes físicas sobre una superficie. Nos enfrentamos entonces a una nueva pregunta: ¿cómo podemos determinar el volumen total atrapado entre el suelo del plano $xy$ y una superficie determinada por una función positiva $z = f(x,y)$?

Para comenzar a responder esto de la manera más sencilla, limitaremos nuestro suelo a una región rectangular del plano, la cual podemos definir matemáticamente como el conjunto de puntos donde la variable $x$ se encuentra atrapada entre los valores $a$ y $b$, mientras que la variable $y$ varía entre los valores $c$ y $d$:

$$R = \{(x,y) : a \leq x \leq b, \ c \leq y \leq d\}$$

Antes de proponer una definición rigurosa que formalice este concepto, es conveniente apoyarnos en la intuición geométrica y en métodos históricos que simplificaron este problema antes de la existencia del cálculo moderno.

## 3. El Principio de Cavalieri

Imagina un cuerpo sólido cualquiera suspendido en el espacio. Si cortamos este objeto con un plano horizontal a una distancia determinada de una base de referencia, la superficie expuesta en el corte se conoce como sección transversal. El matemático italiano Bonaventura Cavalieri postuló que si conocemos el área de estas secciones transversales para cada punto del recorrido del sólido, podemos determinar su volumen total mediante un proceso de acumulación continua.



Formalmente, si llamamos $A(x)$ al área de la sección transversal medida a una distancia $x$ desde el origen, el volumen del cuerpo está dado por la integral en una variable:

$$V = \int_a^b A(x) \, dx$$

Donde $a$ y $b$ representan las distancias mínima y máxima del sólido a lo largo de ese eje de corte. Intuitivamente, si dividiéramos ese intervalo en pequeñas partes de ancho constante, la suma de las áreas de esas rebanadas multiplicadas por su grosor nos daría una excelente aproximación del volumen del cuerpo.

## 4. De las secciones transversales a las integrales iteradas

El Principio de Cavalieri nos ofrece un puente perfecto para calcular el volumen bajo una superficie $z = f(x,y)$ definida en nuestro suelo rectangular. Supongamos que decidimos rebanar este sólido utilizando un plano vertical fijo en una posición particular del eje horizontal, digamos $x = x_0$.

Al hacer esto, la sección transversal que queda expuesta es una región plana cuya altura varía únicamente en función de $y$, desde el límite inferior $y = c$ hasta el límite superior $y = d$. Por lo tanto, el área de esa rebanada en particular se puede calcular mediante una integral clásica de Cálculo I:

$$A(x_0) = \int_c^d f(x_0, y) \, dy$$

Como esta lógica se aplica para cualquier plano de corte a lo largo del eje $x$, podemos sustituir esta expresión dentro de la fórmula general de Cavalieri. El resultado es una estructura matemática que llamaremos **integral iterada**:

$$V = \int_a^b A(x) \, dx = \int_a^b \left[ \int_c^d f(x,y) \, dy \right] dx$$

Este orden de integración nos indica que primero fijamos la variable $x$ como si fuera una constante para integrar respecto a $y$ (la integral interna), y posteriormente integramos el resultado respecto a $x$ (la integral externa). Por supuesto, el proceso es simétrico: podríamos haber realizado los cortes perpendiculares al eje $y$, obteniendo el mismo volumen mediante el orden inverso.

## 5. Definición e ideas geométricas de la Integral Doble

Aunque las integrales iteradas nos ofrecen un método práctico de cálculo, la teoría de la integración requiere una definición que no dependa del orden en que decidamos hacer los cortes. Para ello, extendemos la idea de las sumas de aproximación a dos dimensiones.

En lugar de pensar en una doble sumatoria con índices abstractos que dificultan la lectura, intentemos visualizar el proceso geométricamente. Si queremos estimar el volumen atrapado debajo de una superficie, el camino más natural consiste en dividir el suelo rectangular en una rejilla de cuadritos pequeños. Cada uno de estos cuadritos posee un área elemental idéntica que llamaremos diferencial de área ($\Delta A$).

Sobre cada porción de la rejilla, levantamos un prisma rectangular, o una caja vertical, cuya altura llega exactamente hasta tocar la superficie de la función en algún punto de ese cuadrito. El volumen de una sola de estas cajas se calcula de forma directa multiplicando la altura de la función por el área de su base:

$$\text{Volumen de la caja} = f(x,y) \cdot \Delta A$$

Si sumamos el volumen de todas las cajas que cubren el suelo, obtenemos una estimación numérica del volumen total. Si refinamos la rejilla aumentando la cantidad de divisiones, las cajas se vuelven tan delgadas que sus extremos superiores se adaptan perfectamente a la curvatura de la superficie.



En el límite, cuando las dimensiones de estas bases tienden a cero y el número de cajas tiende a infinito, la aproximación se vuelve exacta y la suma se transforma en la **Integral Doble**, la cual denotamos como:

$$\iint_R f(x,y) \, dA = \iint_R f(x,y) \, dx dy$$

Una de las conclusiones más importantes de este desarrollo teórico es el **Teorema de Fubini**. Este teorema establece que si una función es continua en una región rectangular, el orden en el que realicemos las integrales iteradas no altera el resultado final. Es decir, la integral doble es única y se puede resolver en la dirección que resulte matemáticamente más cómoda:

$$\iint_R f(x,y) \, dA = \int_a^b \int_c^d f(x,y) \, dy \, dx = \int_c^d \int_a^b f(x,y) \, dx \, dy$$

## 6. Ejemplos resueltos analíticamente

Para consolidar estos conceptos, analicemos el desarrollo de dos ejercicios resueltos paso a paso empleando el Teorema de Fubini.

### Ejemplo 1
Evalúe la integral $\iint_R (x^2 + y^2) \, dA$ sobre el rectángulo $R = [-1,1] \times [0,1]$.

**Solución:**
Decidimos estructurar la integral iterada realizando primero la integración respecto a la variable $y$:
$$\int_{-1}^1 \left[ \int_0^1 (x^2 + y^2) \, dy \right] dx$$

Al resolver la integral interna, tratamos a la variable $x$ como una constante:
$$\int_0^1 (x^2 + y^2) \, dy = \left[ x^2y + \frac{y^3}{3} \right]_0^1 = \left( x^2(1) + \frac{1^3}{3} \right) - (0) = x^2 + \frac{1}{3}$$

Ahora, integramos este resultado intermedio respecto a la variable restante, $x$:
$$\int_{-1}^1 \left( x^2 + \frac{1}{3} \right) dx = \left[ \frac{x^3}{3} + \frac{1}{3}x \right]_{-1}^1$$

Evaluando los límites de integración obtenemos el valor final del volumen:
$$\left( \frac{1^3}{3} + \frac{1}{3}(1) \right) - \left( \frac{(-1)^3}{3} + \frac{1}{3}(-1) \right) = \left( \frac{1}{3} + \frac{1}{3} \right) - \left( -\frac{1}{3} - \frac{1}{3} \right) = \frac{2}{3} - \left( -\frac{2}{3} \right) = \frac{4}{3}$$

### Ejemplo 2
Calcule la integral $\iint_S \cos x \sin y \, dA$ donde $S$ es el cuadrado determinado por la región $[0, \pi/2] \times [0, \pi/2]$.

**Solución:**
Planteamos las integrales iteradas en el orden $dx dy$:
$$\int_0^{\pi/2} \left[ \int_0^{\pi/2} \cos x \sin y \, dx \right] dy$$

Integramos primero respecto a $x$, manteniendo a $\sin y$ constante durante este paso:
$$\int_0^{\pi/2} \cos x \sin y \, dx = \sin y \left[ \sin x \right]_0^{\pi/2} = \sin y \left( \sin(\pi/2) - \sin(0) \right) = \sin y (1 - 0) = \sin y$$

Finalmente, integramos el resultado respecto a la variable $y$:
$$\int_0^{\pi/2} \sin y \, dy = \left[ -\cos y \right]_0^{\pi/2} = -\cos(\pi/2) - (-\cos(0)) = 0 - (-1) = 1$$

## 7. Laboratorio computacional: Visualización avanzada y aproximación con Python

Como ingenieros, es fundamental conectar los resultados analíticos que obtenemos en el papel con herramientas de computación gráfica y cálculo numérico. En esta sección utilizaremos la biblioteca `matplotlib` para inspeccionar la geometría de las superficies en tres dimensiones y diseñaremos un pequeño algoritmo para experimentar de forma autónoma con el concepto del límite.

La siguiente celda genera tres representaciones visuales distintas para una superficie: un gráfico de contorno o curvas de nivel, un gráfico de malla de alambre (*wireframe*) y una representación de superficie sólida coloreada.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# Definición de la función de prueba
def f(x, y):
    return x**2 + y**2

# Construcción de la malla de puntos en el plano xy
x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# 1. Gráfico de curvas de nivel (Contornos)
fig1 = plt.figure()
ax1 = plt.axes(projection='3d')
ax1.contour3D(X, Y, Z, 50, cmap='binary')
ax1.set_title('Grafico de curvas de nivel (Contour)')
ax1.set_xlabel('Eje X')
ax1.set_ylabel('Eje Y')
ax1.set_zlabel('Eje Z')

# 2. Gráfico de malla de alambre (Wireframe)
fig2 = plt.figure()
ax2 = plt.axes(projection='3d')
ax2.plot_wireframe(X, Y, Z, color='black')
ax2.set_title('Grafico de malla de alambre (Wireframe)')
ax2.set_xlabel('Eje X')
ax2.set_ylabel('Eje Y')
ax2.set_zlabel('Eje Z')

# 3. Gráfico de superficie sólida (Surface)
fig3 = plt.figure()
ax3 = plt.axes(projection='3d')
ax3.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap='viridis', edgecolor='none')
ax3.set_title('Grafico de superficie solida (Surface)')
ax3.set_xlabel('Eje X')
ax3.set_ylabel('Eje Y')
ax3.set_zlabel('Eje Z')

plt.show()
```

## 🔍 Espacio de Exploración Autónoma: El límite numérico

A continuación dispones de un código diseñado para emular el proceso de aproximación por cajas que discutimos teóricamente. El programa toma la función del Ejemplo 1 ($z = x^2 + y^2$) en la región $R = [-1,1] \times [0,1]$ y calcula el volumen sumando los prismas rectangulares de la rejilla.

En este momento, el código está configurado para usar únicamente **4 cajas en total** (2 en el eje X y 2 en el eje Y). Como recordarás, el valor real analítico exacto de esta integral es $4/3 \approx 1.333333$.

**Tu reto autónomo:**
1. Ejecuta la celda tal como está. Analiza la magnitud del error de cálculo numérico obtenido con una rejilla tan pequeña.
2. Modifica de forma autónoma las variables `cajas_x` y `cajas_y`. Prueba configurando la rejilla para 10, luego para 50, y finalmente para 100 cajas por eje.
3. Observa el comportamiento del error. Escribe una breve reflexión en tu bitácora de clase sobre cómo la computadora manifiesta el concepto matemático de "tender al infinito" a través de este bucle numérico.

```python
import numpy as np

# Definición de la función de altura y los límites del rectángulo
def f(x, y):
    return x**2 + y**2

x_min, x_max = -1.0, 1.0
y_min, y_max = 0.0, 1.0

# ====================================================
# EXPERIMENTO: MODIFICA ESTOS DOS VALORES:
cajas_x = 2
cajas_y = 2
# ====================================================

# Cálculo de las dimensions de la base de cada caja
dx = (x_max - x_min) / cajas_x
dy = (y_max - y_min) / cajas_y
delta_A = dx * dy

# Determinación de los puntos medios para evaluar la altura en cada celda
puntos_x = np.linspace(x_min + dx/2, x_max - dx/2, cajas_x)
puntos_y = np.linspace(y_min + dy/2, y_max - dy/2, cajas_y)

# Algoritmo de acumulación de volúmenes de las cajas
volumen_aproximado = 0.0
for x in puntos_x:
    for y in puntos_y:
        altura_caja = f(x, y)
        volumen_aproximado += altura_caja * delta_A

# Evaluación de la precisión frente al valor teórico exacto
volumen_real = 4/3
error = abs(volumen_real - volumen_aproximado)

print(f"--- RESULTADOS CON REJILLA DE {cajas_x}x{cajas_y} ({cajas_x * cajas_y} CAJAS) ---")
print(f"Volumen aproximado calculado: {volumen_aproximado:.6f}")
print(f"Volumen real analítico:      {volumen_real:.6f}")
print(f"Error de aproximación:        {error:.6f}")
```
