### **1. Más allá de los rectángulos: Regiones generales en el plano**

En nuestra sesión anterior exploramos cómo calcular el volumen atrapado bajo una superficie cuando el suelo o dominio de integración era un rectángulo perfecto. Sin embargo, en la práctica de la ingeniería, los elementos estructurales, las láminas mecánicas o los campos de distribución rara vez tienen formas tan simétricas; por lo general, están delimitados por curvas, rectas o geometrías complejas.

Para abordar este problema, el cálculo multivariable clasifica las regiones del suelo en dos grandes categorías que facilitan su estudio: regiones simples respecto al eje Y (regiones $y$-simples) y regiones simples respecto al eje X (regiones $x$-simples).

Decimos que una región es **Y- simple** si sus límites laterales izquierdo y derecho están determinados por dos valores fijos y rectos ($x = a$ y $x = b$), mientras que sus límites superior e inferior están definidos por curvas que dependen de la variable $x$. En este tipo de regiones, si trazamos una línea vertical imaginaria que cruce el suelo, esta siempre entrará por una curva inferior conocida y saldrá por una curva superior predecible.



Por otro lado, una región  **X- simple** si ocurre lo contrario. Sus límites inferior y superior están determinados por líneas horizontales fijas ($y = c$ y $y = d$), mientras que sus costados izquierdo y derecho están limitados por curvas que varían en función de la altura $y$. Si trazamos una línea horizontal que cruce esta región, entrará por una curva izquierda y saldrá por una curva derecha perfectamente identificables.

### **2. ¿Cómo se define formalmente la integral en estas regiones?**

Para unificar los criterios matemáticos sin enredarnos en demostraciones abstractas, imagina que tomamos una región del suelo con bordes curvos y la encerramos por completo dentro de una caja rectangular ficticia mucho más grande, a la que llamaremos $R$.

Para poder integrar nuestra función original sobre esa nueva región rectangular regular, hacemos un truco matemático muy sencillo: redefinimos la función de tal manera que conserve sus valores reales y sus alturas originales si el punto medido está dentro de la zona curva, pero le asignamos una altura de cero a cualquier punto que quede fuera de la zona curva (en el espacio sobrante del rectángulo).

Al calcular la integral doble sobre todo el rectángulo con esta función modificada, el área sobrante no aporta nada al volumen final. De esta forma tan ingeniosa, reducimos el problema de una región curva al de una región rectangular que ya sabemos resolver mediante el Teorema de Fubini.

---

### **3. Reducción a integrales iteradas**

Este planteamiento teórico nos permite deducir cómo calcular las integrales en la práctica mediante integrales sucesivas. Dependiendo de la geometría del suelo, utilizaremos uno de los siguientes enfoques de resolución:

Si el suelo es una región simple respecto al eje Y, la variable $x$ se mueve entre los números fijos $a$ y $b$, mientras que $y$ varía entre las funciones curvas $\phi_1(x)$ y $\phi_2(x)$. El volumen se calcula resolviendo primero la integral interna respecto a $y$:

$$\iint_D f(x,y) \, dA = \int_a^b \int_{\phi_1(x)}^{\phi_2(x)} f(x,y) \, dy \, dx$$

Si el suelo es una región simple respecto al eje X, la variable $y$ se encuentra atrapada entre los valores fijos $c$ y $d$, mientras que $x$ se mueve horizontalmente entre las curvas $\psi_1(y)$ y $\psi_2(y)$. En este escenario, resolvemos primero la integral interna respecto a $x$:

$$\iint_D f(x,y) \, dA = \int_c^d \int_{\psi_1(y)}^{\psi_2(y)} f(x,y) \, dx \, dy$$

*Nota clave para el análisis:* Las variables que corresponden a límites de integración curvos (funciones) siempre deben resolverse en la integral interna, dejando los valores numéricos fijos y constantes para la última integral externa. El resultado de una integral doble siempre debe ser un número que represente una magnitud física real.

### **4. Ejemplos resueltos paso a paso**

Estudiemos con detenimiento el planteamiento analítico de dos situaciones típicas en ingeniería.

#### **Ejemplo 1**
Determine el valor de la integral $\iint_T (x^3 y + \cos x) \, dA$, donde el dominio de integración $T$ es un triángulo delimitado por los puntos del suelo donde $0 \leq x \leq \pi/2$ y $0 \leq y \leq x$.

**Solución:**
Observamos que los límites para $x$ son constantes numéricas fijas, mientras que el límite superior de $y$ varía linealmente según el valor de $x$. Planteamos la integral iterada colocando a $y$ en la etapa interna:

$$\int_0^{\pi/2} \left[ \int_0^x (x^3 y + \cos x) \, dy \right] dx$$

1. Resolvemos la integral interna respecto a $y$ tratando a $x$ como una constante:
   $$\int_0^x (x^3 y + \cos x) \, dy = \left[ x^3 \frac{y^2}{2} + y \cos x \right]_0^x$$
   $$\text{Evaluando en el límite superior } y = x: \quad \frac{x^5}{2} + x \cos x$$
   $$\text{Evaluando en el límite inferior } y = 0: \quad 0 + 0 = 0$$

2. Ahora, integramos la expresión resultante respecto a $x$ desde $0$ hasta $\pi/2$:
   $$\int_0^{\pi/2} \left( \frac{x^5}{2} + x \cos x \right) dx = \left[ \frac{x^6}{12} \right]_0^{\pi/2} + \int_0^{\pi/2} x \cos x \, dx$$

   Para el término de la derecha aplicamos integración por partes ($\int u \, dv = uv - \int v \, du$), seleccionando $u = x$ y $dv = \cos x \, dx$, lo que nos da $du = dx$ y $v = \sin x$:
   $$\int_0^{\pi/2} x \cos x \, dx = \left[ x \sin x \right]_0^{\pi/2} - \int_0^{\pi/2} \sin x \, dx = \left[ x \sin x + \cos x \right]_0^{\pi/2}$$

   Uniendo ambas partes y evaluando en los límites extremos de la región obtenemos:
   $$\left[ \frac{x^6}{12} + x \sin x + \cos x \right]_0^{\pi/2} = \left( \frac{(\pi/2)^6}{12} + \frac{\pi}{2}\sin(\pi/2) + \cos(\pi/2) \right) - \left( 0 + 0 + \cos(0) \right)$$
   $$\text{Resultado analítico exacto: } \frac{\pi^6}{768} + \frac{\pi}{2} - 1$$

Calcule el volumen del tetraedro en el primer octante limitado por los planos coordenados $x=0$, $y=0$, $z=0$ y el plano inclinado $y - x + z = 1$.

**Solución:**

El cuerpo está limitado superiormente por el plano inclinado, por lo que despejamos la altura del sólido: $z = 1 + x - y$. Para identificar la región del suelo, miramos qué pasa cuando el plano toca el piso (haciendo $z=0$), lo que nos da la recta límite $y = 1 + x$, o visto de otra forma, $x = y - 1$.

Si decidimos analizar el suelo como una región simple respecto al eje X en el primer octante, vemos que la altura vertical $y$ varía entre los valores fijos $0$ y $1$. Para cada altura $y$, la variable horizontal $x$ avanza desde la pared izquierda $x = 0$ hasta encontrarse con la recta límite del plano, que despejada es $x = y - 1$. Sin embargo, dado que estamos trabajando en el primer octante donde las variables son positivas, la frontera natural del triángulo en el suelo nos indica que $x$ varía desde $0$ hasta $1-y$ (si invertimos la orientación del plano base).

Planteando el volumen bajo el enfoque adecuado para este triángulo elemental:
$$\int_0^1 \int_0^{1-y} (1 + x - y) \, dx \, dy$$

1. Integramos primero respecto a la variable horizontal $x$:
   $$\int_0^{1-y} (1 + x - y) \, dx = \left[ x + \frac{x^2}{2} - yx \right]_0^{1-y} = (1-y) + \frac{(1-y)^2}{2} - y(1-y)$$
   $$\text{Simplificando algebraicamente la expresión interna: } \frac{1}{2} - y + \frac{y^2}{2}$$

2. Integramos el resultado respecto a la variable vertical restante $y$:
   $$\int_0^1 \left( \frac{1}{2} - y + \frac{y^2}{2} \right) dy = \left[ \frac{1}{2}y - \frac{y^2}{2} + \frac{y^3}{6} \right]_0^1 = \left( \frac{1}{2} - \frac{1}{2} + \frac{1}{6} \right) - 0 = \frac{1}{6}$$

### **5. Laboratorio computacional: Integración numérica con SciPy**

El cálculo de integrales en regiones con límites curvos o inclinados puede volverse propenso a errores algebraicos cuando se realiza manualmente. Como ingenieros, contamos con la biblioteca científica `SciPy` de Python, la cual provee herramientas robustas para evaluar estas expresiones numéricamente con alta precisión.

A continuación, estudiaremos el uso de la función `dblquad` (diseñada para integrales dobles). A diferencia de las regiones rectangulares, cuando trabajamos con regiones generales, los límites de la integral interna de `dblquad` no se ingresan como números estáticos, sino como funciones matemáticas (`lambda`).

*Regla de SciPy:* La función `dblquad` requiere obligatoriamente que la función de altura reciba los parámetros en el orden inverso, es decir, `lambda y, x`, y asume de forma predeterminada que la integral interna corresponde a la variable $y$ y la externa a la variable $x$.

Ejecuta las siguientes celdas de código para observar cómo se traducen matemáticamente los límites a la sintaxis del lenguaje de programación.

```python
from scipy.integrate import dblquad
from numpy import cos, pi

print("--- Verificación del Ejemplo 1 ---")

# f es la función de altura. Nota que escribimos (y, x) por estándar de dblquad
f_triangulo = lambda y, x: x**3 * y + cos(x)

# Límites externos fijos para x: de 0 a pi/2
lim_x_inf = 0
lim_x_sup = pi/2

# Límites internos para y: son funciones que dependen de x.
# El límite inferior es la función constante cero y el superior es la recta y = x
lim_y_inf = lambda x: 0
lim_y_sup = lambda x: x

# Ejecución del cálculo numérico
resultado, error_estimado = dblquad(f_triangulo, lim_x_inf, lim_x_sup, lim_y_inf, lim_y_sup)

print(f"Resultado numérico de la integral: {resultado:.6f}")
print(f"Margen de error estimado por SciPy:  {error_estimado:.1e}")

# Valor analítico para comparar: (pi**6)/768 + pi/2 - 1
valor_exacto = (pi**6)/768 + (pi/2) - 1
print(f"Valor analítico teórico exacto:   {valor_exacto:.6f}")
```

#### **Uso avanzado con la función nquad**

Cuando las regiones tienen definiciones donde es más cómodo declarar las fronteras de forma explícita e independiente, podemos recurrir a `nquad`. Esta función es más versátil y nos permite separar claramente los límites de integración mediante funciones tradicionales de Python, lo que resulta ideal para documentar códigos de ingeniería más legibles.

```python
from scipy import integrate

print("--- Cálculo alternativo usando nquad ---")

# Función de altura para un campo de densidades f(x,y) = x * y
def funcion_densidad(x, y):
    return x * y

# Definición de las fronteras de la región mediante funciones independientes
def limites_y():
    return [0, 0.5]

def limites_x(y):
    return [0, 1 - 2*y]

# nquad recibe una lista con las funciones de los límites en orden inverso de integración
resultado_campo, error_campo = integrate.nquad(funcion_densidad, [limites_x, limites_y])

print(f"Resultado del cálculo del campo: {resultado_campo:.6f}")
```

### **6. El cambio en el orden de integración**

Existen ocasiones donde una integral iterada planteada en un orden específico (por ejemplo, $dy dx$) resulta sumamente difícil o incluso imposible de resolver mediante técnicas analíticas tradicionales (como sustitución o partes), debido a la naturaleza de la función interna.

Si la región del suelo es regular y cumple con las condiciones de ser tanto $x$-simple como $y$-simple al mismo tiempo, el Teorema de Fubini nos permite realizar una operación estratégica: cambiar el orden de integración.

Cambiar el orden **no consiste simplemente en cambiar la posición de los símbolos $dx$ y $dy$** ni en intercambiar los límites numéricos de lugar. El proceso requiere dibujar la región original del suelo, analizar sus fronteras desde la perspectiva del otro eje coordenado y construir los nuevos límites de integración correspondientes. Esta técnica es una herramienta invaluable en la caja de herramientas de cualquier ingeniero para simplificar problemas de cálculo computacional y analítico.

### **7. Ejemplo de reconfiguración de límites**

Evaluemos el impacto de esta técnica analizando la siguiente integral:
$$\int_1^2 \int_0^{\log(x)} (x-1)\sqrt{1+e^{2y}} \, dy \, dx$$

Si intentamos resolver esta expresión de forma directa respetando el orden original, nos topamos con la integral interna respecto a $y$: $\int \sqrt{1+e^{2y}} \, dy$. Esta integral requiere una sustitución trigonométrica o hiperbólica compleja que alarga significativamente el desarrollo del problema.

**Análisis para el cambio de orden:**
1. Al examinar los límites originales, vemos que el suelo está definido por las líneas verticales fijos $x=1$ y $x=2$. En la dirección vertical, la región está acotada inferiormente por el eje horizontal $y=0$ y superiormente por la curva logarítmica $y = \log(x)$.
2. Si despejamos la variable horizontal en la ecuación de la curva para cambiar la perspectiva, la relación inversa es $x = e^y$.
3. Al mirar la región desde el eje Y, determinamos los nuevos límites fijos globales: el valor mínimo de la altura es $y = 0$, y el valor máximo ocurre cuando la curva llega al extremo derecho del intervalo ($x=2$), es decir, $y = \log(2)$.
4. Para cualquier altura intermedia $y$, la variable horizontal $x$ entra a la región por la curva izquierda $x = e^y$ y sale por la pared recta derecha $x = 2$.

Aplicando el Teorema de Fubini con la nueva configuración del suelo, la integral equivalente en el orden inverso ($dx dy$) es:
$$\int_0^{\log(2)} \int_{e^y}^2 (x-1)\sqrt{1+e^{2y}} \, dx \, dy$$

Notemos la enorme ventaja: la nueva integral interna se resuelve respecto a $x$, por lo que el término de la raíz con la función exponencial se comporta como una simple constante temporal. La integración se vuelve puramente polinomial y directa.

```python
from numpy import sqrt, log, exp
from scipy.integrate import dblquad

print("--- EXPERIMENTO: EVALUACIÓN DE ÓRDENES DE INTEGRACIÓN ---")

# 1. Evaluación en el orden original (dy dx)
# Límites: x va de 1 a 2; y va de 0 a log(x)
f_orden_1 = lambda y, x: (x - 1) * sqrt(1 + exp(2 * y))
volumen_orden_1, _ = dblquad(f_orden_1, 1, 2, lambda x: 0, lambda x: log(x))

# 2. Evaluación en el orden invertido (dx dy) después del análisis geométrico
# Límites: y va de 0 a log(2); x va de exp(y) a 2
f_orden_2 = lambda x, y: (x - 1) * sqrt(1 + exp(2 * y))
volumen_orden_2, _ = dblquad(f_orden_2, 0, log(2), lambda y: exp(y), lambda y: 2)

print(f"Calculo en el orden original (dy dx) : {volumen_orden_1:.6f}")
print(f"Calculo en el orden cambiado (dx dy) : {volumen_orden_2:.6f}")
print(f"Diferencia numérica entre ambos      : {abs(volumen_orden_1 - volumen_orden_2):.1e}")
```
