### **1. Introducción a las Integrales Triples**

En los capítulos anteriores aprendimos a calcular volúmenes extendiendo el concepto de integración a dos dimensiones sobre una región plana del suelo. Ahora daremos el siguiente paso lógico en el cálculo multivariable: integrar funciones sobre regiones tridimensionales sólidas del espacio.

Imagina un cuerpo sólido en el espacio,  o un bloque tridimensional al que llamaremos $B$. Si este bloque es una caja rectangular perfecta definida por los intervalos de sus tres ejes principales, $B = [a, b] \times [c, d] \times [p, q]$, podemos aproximar cualquier propiedad acumulativa del sólido (como su masa total) dividiéndolo en una cuadrícula tridimensional de pequeños paralelepípedos elementales.

Cada uno de estos pequeños cubos tiene un volumen elemental denotado por $\Delta V = \Delta x \Delta y \Delta z$. Al evaluar la función en un punto interior de cada cubo y sumar las contribuciones de todos ellos, construimos una suma triple de Riemann:

$$S_n = \sum_{i=0}^{n-1} \sum_{j=0}^{n-1} \sum_{k=0}^{n-1} f(c_{ijk}) \Delta V$$

Si la función es continua y acotada, al hacer la subdivisión infinitamente fina ($n \to \infty$), esta suma converge a un valor único que llamamos la **integral triple** de $f$ sobre el sólido $B$. La denotamos indistintamente como:

$$\iiint_B f(x,y,z) \, dV \quad \text{o} \quad \iiint_B f(x,y,z) \, dx \, dy \, dz$$

### **2. El Teorema de Fubini en Tres Dimensiones**

Cuando el dominio de integración es un bloque rectangular tridimensional rígido (una caja), la generalización del Teorema de Fubini nos indica que podemos calcular la integral triple mediante tres integraciones sucesivas (iteradas).

Dado que todos los límites en este escenario son números constantes fijos, el orden en el que elijas colocar los diferenciales ($dx$, $dy$, $dz$) no alterará en lo absoluto el resultado final. Tienes 6 permutaciones posibles para jugar:

$$\iiint_B f(x,y,z) \, dV = \int_p^q \int_c^d \int_a^b f(x,y,z) \, dx \, dy \, dz = \int_c^d \int_a^b \int_p^q f(x,y,z) \, dz \, dx \, dy$$

### **4. Regiones Elementales en el Espacio**

En los problemas de ingeniería, los sólidos raramente son cubos perfectos. Para integrarlos de manera ordenada, clasificamos los dominios en **regiones tridimensionales elementales**.

Una región es elemental si podemos "atrapar" una de las variables (por lo general la altura $z$) entre dos sábanas o superficies continuas que dependen de las variables restantes: $\gamma_1(x,y) \leq z \leq \gamma_2(x,y)$. A su vez, la sombra o proyección de ese sólido sobre el piso debe corresponder a una región bidimensional simple ($x$-simple o $y$-simple).



Esto nos permite estructurar la reducción a integrales iteradas bajo dos esquemas de límites variables:

$$\iiint_W f(x,y,z) \, dV = \int_a^b \int_{\phi_1(x)}^{\phi_2(x)} \int_{\gamma_1(x,y)}^{\gamma_2(x,y)} f(x,y,z) \, dz \, dy \, dx$$

$$\iiint_W f(x,y,z) \, dV = \int_c^d \int_{\psi_1(y)}^{\psi_2(y)} \int_{\gamma_1(x,y)}^{\gamma_2(x,y)} f(x,y,z) \, dz \, dx \, dy$$

*Observación:* Las funciones con mayor número de variables independientes siempre se resuelven primero en las etapas más internas de la integral, migrando gradualmente hacia funciones de una sola variable, hasta concluir con números constantes puros en la última frontera exterior.

### **5. Laboratorio Computacional: Integración Triple con SciPy**

Para evaluar numéricamente integrales en tres dimensiones con fronteras curvas, la biblioteca `SciPy` nos proporciona la función especializada `tplquad`.

*Regla estructural de entrada:* Al igual que ocurría con las integrales dobles, la función matemática de altura debe declararse siguiendo el orden inverso de las variables en los argumentos del lenguaje, es decir, `lambda z, y, x`. Además, los límites de integración se suministran de afuera hacia adentro: primero las constantes fijas de la variable externa $x$, luego las funciones de la frontera intermedia $y(x)$, y finalmente las funciones de la frontera interna $z(x,y)$.

Estudia detenidamente el código de la siguiente celda para comprender cómo se mapean los límites matemáticos del Ejemplo 1 y del Ejemplo 3 en la sintaxis de Python.

```python
from scipy.integrate import tplquad
import numpy as np

print("--- Validación Numérica de Ejemplos ---")

# =====================================================================
# CASO A: Evaluación del Cubo Unitario (Ejemplo 1)
# Función f(z, y, x) = (x + 2y + 3z)^2
# =====================================================================
f_cubo = lambda z, y, x: (x + 2*y + 3*z)**2

# Límites constantes del cubo para el experimento numérico
res_cubo, err_cubo = tplquad(f_cubo, 0, 1, lambda x: 0, lambda x: 1, lambda x, y: 0, lambda x, y: 1)
print(f"1. Resultado de la integral sobre el cubo: {res_cubo:.6f}")


# =====================================================================
# CASO B: Evaluación del Sólido Parabólico W (Ejemplo 3)
# Función f(z, y, x) = x
# Límites: x va de 0 a sqrt(2)
#          y va de 0 a la proyección del cilindro en el suelo: sqrt(2 - x^2)
#          z va desde el paraboloide (x^2 + y^2) hasta el techo plano z = 2
# =====================================================================
f_paraboloide = lambda z, y, x: x

lim_x_inf = 0
lim_x_sup = np.sqrt(2)

lim_y_inf = lambda x: 0
lim_y_sup = lambda x: np.sqrt(2 - x**2)

lim_z_inf = lambda x, y: x**2 + y**2
lim_z_sup = lambda x, y: 2

res_solido, err_solido = tplquad(f_paraboloide, lim_x_inf, lim_x_sup,
                                 lim_y_inf, lim_y_sup,
                                 lim_z_inf, lim_z_sup)

print(f"2. Resultado de la integral sobre la región W: {res_solido:.6f}")
```

### **6. Geometría de las Funciones de $\mathbb{R}^2 \to \mathbb{R}^2$ (Mapeos)**

Cuando integramos sobre regiones con simetrías circulares, elípticas o cilíndricas, el uso de las coordenadas rectangulares tradicionales ($x,y,z$) suele dar lugar a raíces cuadradas complejas en los límites de integración que dificultan el cálculo numérico y analítico. Para solucionar esto, recurrimos al concepto de **Transformación o Mapeo**.

Un mapeo $T$ es una función que toma puntos de un espacio de entrada (por ejemplo, un plano con coordenadas $u$ y $v$) y los estira, dobla o deforma de manera continua para dar origen a una nueva geometría en un espacio de salida (con coordenadas $x$ e $y$). Denotamos esta correspondencia como $D = T(D^*)$.

Un ejemplo clásico de ingeniería es la transformación a **Coordenadas Polares**, definida por las ecuaciones:
$$T(r, \theta) = (r \cos\theta, r \sin\theta)$$



Este mapeo tiene una propiedad geométrica asombrosa: toma una región completamente recta y rectangular en el plano de entrada $r\theta$ (donde el radio avanza de $0$ a $R$ y el ángulo avanza de $0$ a $2\pi$) y la curva suavemente en el espacio de salida hasta transformarla en un disco o círculo perfecto de radio $R$.

### **7. El Teorema de Cambio de Variable y el Determinante Jacobiano**

Si deseamos trasladar una integral doble o triple desde una región complicada $D$ hacia un bloque ordenado y rectangular $D^*$, no basta con sustituir las ecuaciones de la transformación dentro de la función original. Al estirar o curvar el espacio, las áreas de los pequeños cuadrados elementales cambian de tamaño.

Para corregir esta distorsión en la escala del espacio, debemos multiplicar el integrando por un factor de compensación geométrica denominado **Determinante Jacobiano**. El Jacobiano mide la tasa de cambio del área o volumen transformada, y se calcula como el determinante de la matriz de derivadas parciales del mapeo:

$$\frac{\partial(x,y)}{\partial(u,v)} = \begin{vmatrix} \frac{\partial x}{\partial u} & \frac{\partial x}{\partial v} \\ \frac{\partial y}{\partial u} & \frac{\partial y}{\partial v} \end{vmatrix}$$

Al incorporar este factor de escala, el Teorema del Cambio de Variable para integrales dobles se enuncia formalmente como:

$$\iint_D f(x,y) \, dx \, dy = \iint_{D^*} f(x(u,v), y(u,v)) \left| \frac{\partial(x,y)}{\partial(u,v)} \right| \, du \, dv$$

Si la transformación no distorsiona el área (es decir, si el módulo del Jacobiano es igual a 1), el tamaño de las regiones se conserva idéntico. Un ejemplo de esto ocurre con las transformaciones lineales puras que convierten paralelogramos en cuadrados sin alterar la homogeneidad del espacio.

### **8. Extensiones a Tres Dimensiones: Bloques Cilíndricos y Esféricos**

Cuando extendemos este principio al espacio tridimensional, el determinante jacobiano se convierte en una matriz de $3 \times 3$. Estudiemos las dos transformaciones más utilizadas en la ingeniería física para modelar campos de fuerzas o fluidos:

#### **Coordenadas Cilíndricas**
Modela el espacio usando cilindros concéntricos alrededor del eje vertical Z:
$$x = r \cos\theta, \quad y = r \sin\theta, \quad z = z$$

*Factor de volumen corregido:* $dV = r \, dz \, dr \, d\theta$ (El Jacobiano de esta transformación es simplemente el radio $r$).

#### **Coordenadas Esféricas**
Modela el espacio mediante esferas concéntricas midiendo el radio de inclinación $\rho$, el ángulo azimutal $\theta$ en el suelo, y el ángulo polar $\phi$ desde el eje vertical:
$$x = \rho \sin\phi \cos\theta, \quad y = \rho \sin\phi \sin\theta, \quad z = \rho \cos\phi$$

*Factor de volumen corregido:* $dV = \rho^2 \sin\phi \, d\rho \, d\phi \, d\theta$.

### **9. Laboratorio de Experimentación Autónoma: El Impacto Numérico del Cambio de Variable**

Consideremos un paralelogramo plano $P$ delimitado en el suelo por las siguientes cuatro líneas rectas cruzadas:
$$y = 2x, \quad y = 2x - 2, \quad y = x, \quad y = x + 1$$

Evaluar de manera directa la integral $\iint_P x y \, dx \, dy$ requeriría partir la región en tres sub-triángulos independientes debido a los quiebres de las esquinas. Para simplificar el problema, proponemos un cambio de variables lineal diseñado a medida para este sólido:
$$x(u,v) = u - v, \quad y(u,v) = 2u - v$$

**Tu reto en este laboratorio:**
1. Calcula analíticamente en tu cuaderno el determinante Jacobiano asociado a esta transformación lineal $x(u,v)$ e $y(u,v)$. Deberías obtener un valor constante igual a $1$.
2. Ejecuta la siguiente celda de código. El programa evaluará de forma automática los límites transformados de la nueva región regular en el espacio de entrada $uv$, la cual se reduce a un rectángulo simple donde $0 \leq u \leq 1$ y $-1 \leq v \leq 0$.
3. Analiza los resultados impresos. Observa cómo la computadora resuelve ambas estructuras de integración y reflexiona sobre la ventaja de realizar transformaciones espaciales en problemas avanzados de ingeniería.

```python
from scipy.integrate import dblquad

print("--- EVALUACIÓN COMPUTACIONAL DEL CAMBIO DE VARIABLE ---")

# 1. Definimos la función original expresada en las nuevas coordenadas (u, v)
# f(x, y) = x * y  -->  f(u, v) = (u - v) * (2u - v)
# Multiplicamos por el valor absoluto del Jacobiano calculado (que es 1)
f_transformada = lambda v, u: (u - v) * (2*u - v) * 1

# 2. Los límites de la nueva región rectangular D* son constantes puras:
# u varía entre 0 y 1
# v varía entre -1 y 0
lim_u_inf = 0
lim_u_sup = 1
lim_v_inf = lambda u: -1
lim_v_sup = lambda u: 0

# 3. Cálculo de la integral sobre la región simplificada
resultado_mapeo, err_mapeo = dblquad(f_transformada, lim_u_inf, lim_u_sup, lim_v_inf, lim_v_sup)

print(f"Valor de la integral usando la transformación lineal: {resultado_mapeo:.6f}")
print(f"Confianza del cálculo numérico (error estimado):     {err_mapeo:.1e}")
```
