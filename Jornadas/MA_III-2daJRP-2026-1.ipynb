## Subespacios fundamentales de una matriz

Considere el sistema de ecuaciones $A\mathbf{x}=\mathbf{b}$, donde

$$
A=
\begin{bmatrix}
1 & 0 & -1\\
0 & 1 & -1\\
1 & 1 & -2
\end{bmatrix},
\quad
\mathbf{x}=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix},
\quad
\mathbf{b}=
\begin{bmatrix}
1\\
1\\
1
\end{bmatrix}.
$$

a. Considerando la matriz $A$, responda las siguientes preguntas: ¿Pueden las columnas o las filas de $A$ constituir una base de $\mathbb{R}^3$?  ¿Pertenece el vector $\mathbf{b}$ al espacio columna de $A$? Justifique cuidadosamente cada una de sus respuestas.

b. Encuentre una base del espacio fila, es decir, del subespacio $C(A^T)$ asociado a la matriz $A$, y determine su dimensión. c. Diseñe un applet en GeoGebra que permita representar geométricamente los subespacios obtenidos en el punto anterior.

d. Utilizando el algoritmo de Gram-Schmidt construido en la entrega VI, halle una base ortonormal del espacio $C(A^T)$.

e. A partir del applet de GeoGebra del punto anterior, realice una construcción geométrica que permita descomponer el vector $\mathbf{b}$ como

$$
\mathbf{b}=\mathbf{v}+\mathbf{w},
$$

donde $\mathbf{v}\in C(A^T)$ y $\mathbf{w}\in N(A)$.

Puede apoyarse, si lo considera pertinente, en la base ortonormal obtenida en el punto anterior.
