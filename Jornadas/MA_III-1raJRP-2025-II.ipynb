## Cambio de variable e integrales dobles

1. (a) Considere la región $D^*$ correspondiente al paralelogramo de vertices $(0, 2), (1, 1), (0, 0),$ y $ (-1,1)$. Explique por qué es posible encontrar una transformación lineal que envíe dicho paralelogramo en el rectángulo $D$ de vértices $(2, 2), (0,1), (0,0),$ y $ (2, 1)$. Los vértices son enviados respectivamente.

(b) Encuentre una transformación lineal de la forma $T(\mathbf{x})=A\mathbf{x}$, donde $A$ es una matriz construida a partir de la solución al sistema $$T(x,y)=\begin{bmatrix}
  a & b \\
  c & d
\end{bmatrix}\begin{bmatrix}
  x \\
  y
\end{bmatrix} = \begin{bmatrix}
  u \\
  v
\end{bmatrix}.$$ Explique por qué tiene sentido resolver el sistema anterior para hallar la matriz $A$.

(c) Explique por qué puede usar el punto anterior para hallar una fórmula de cambio de variables apropiada para transformar el paralelogramo $D^*$ en la región $D$.

(d) Usando el cambio de variables apropiado, defina una función $f$ y una función $g$ para evaluar una integral de la forma

$$\int\int_D f(x,y)e^{g(x,y)}\, dx dy,$$
sobre la región $D^*$ en términos de las variables $u$ y $v$.
