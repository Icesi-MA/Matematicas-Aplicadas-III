## Mínimos cuadrados sobre la bola unitaria

Vamos a estudiar el siguiente problema

$$\min_{||x||^2=1} ||A\mathbf{x}-\mathbf{b}||^2\qquad (1)$$

donde $A\in\mathbb{R}^{m\times n}$, $\mathbf{x}\in\mathbb{R}^{n}$ y $\mathbf{b}\in\mathbb{R}^{m}$. La restricción está pensada para evitar oscilaciones fuertes de la función lineal de ajuste. Para resolver el problema (1) utilizaremos una formulación de optimización en varias variables con el método de multiplicadores de Lagrange. 

1. Formule la función lagrangiana asociada al problema (1), utilizando $+$ para añadir el multiplicador de Lagrange y la restricción.
2. Intente mostrar que la solución de este problema viene dada por el sistema

$$(A^TA+\lambda I_n)\mathbf{x}=A^T\mathbf{b}.$$

3. Sea $\mathbf{b}=\begin{bmatrix}
4\\
2\\
3
\end{bmatrix}$ y $A= \begin{bmatrix}
2&0\\
0&1\\
0&0
\end{bmatrix}$. Utilice la descomposición en valores singulares, $A=U\Sigma V^T$,  de la matriz $A$, para encontrar como antes el vector solución:

$$\mathbf{x}=\sum_{i=1}^r\frac{\mathbf{u}_i^T\mathbf{b}}{\sigma_i}\mathbf{v}_i,$$ donde los vectores $\mathbf{u}_i$ y $\mathbf{v}_i$ son los vectores columna de las matries $U$ y $V$ respectivamente, $r$ es el rango de la matriz $A$ y $\sigma_i,\,i=1,2,$ son los valores singulares de la matreiz $A$.. 

4. Resuelva el problema (1) encontrando un $\lambda^*$ tal que $$\sum_{i=1}^r\left(\frac{\sigma_ib_i}{\sigma_i^2+\lambda^*}\right)^2=1.$$  y en ese caso calcule para el $\lambda^*$ encontrado:

$$\mathbf{x^*}=\sum_{i=1}^r\left(\frac{\sigma_ib_i}{\sigma_i^2+\lambda^*}\right)v_i.$$


5. Compare las soluciones  $\mathbf{x}^*$ y $\mathbf{x}$. ¿Ambas soluciones son válidas? ¿Cúal de las dos soluciones es mejor? Argumente sus respuestas.
