1. (10 ptos) **Formulación de sistemas de ecuaciones lineales**

i. *Interpolación de funciones racionales.* Una función racional de grado dos tiene la forma:
$$f(t) =\frac{c_1 + c_2t + c_3t^2}{1 + d_1t + d_2t^2}$$
donde $c_1, c_2, c_3, d_1$ y $ d_2$ son coeficientes. Considere las condiciones de interpolación
$f(t_i) = y_i;\quad i = 1,...,K;$
donde $t_i$ y $y_i$ son números dados. Expresar las condiciones de interpolación como un conjunto de ecuaciones lineales
 con vector de coeficientes  $\mathbf{c} = (c_1, c_2, c_3, d_1, d_2)$, como $A\mathbf{c}  = \mathbf{b}$. Escriba explícitamente la matriz $ A$, el vector $\mathbf{b}$, y sus respectivas dimensiones.

ii. *Mezcla de crudo.* Un conjunto de $K$ diferentes tipos de petróleo crudo se combinan (mezclan) en
proporciones  $\theta_1,...,\theta_K$. La suma de estas proporciones deben sumar uno; también deben ser no negativss, pero
se ignorará ese requisito aquí. Asociado con el petróleo crudo tipo $k$ hay un $n-$vector $c_k$ que da su concentración de $n$ constituyentes diferentes, tales como hidrocarburos específicos. Encontrar un conjunto de ecuaciones lineales sobre los coeficientes de mezcla, $A\mathbf{\theta}  = \mathbf{b}$, para garantizar
que el petróleo crudo mezclado alcance un conjunto objetivo de concentraciones constituyentes, dado por
el $n-$vector $c^{obj}$. (Incluya la condición  $\theta_1 +...+  \theta_K = 1$  en sus ecuaciones).

\end{bmatrix},$$

donde $a$ es un $n-$vector.

a) ¿Indique qué condiciones debe cumplir el vector $a$ para que $A$ sea una matriz invertible?

$$f(t) =\frac{c_1 + c_2t + c_3t^2}{1 + d_1t + d_2t^2}$$

$$f(1) = 2,\quad f(2) = 5,\quad f(3) = 9,\quad f(4) = -1,\quad f(5) = -4.$$
