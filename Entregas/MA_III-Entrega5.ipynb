Puede consultar mas detalles de la factorización $LU$ en cualquier texto de algebra lineal, sugiero los textos de Gilbert Strang y Kolman. Mas abajo encontrará un pequeño ejemplo de como encontrar la factorización $LU$ en python. Debe usarla para contestar y argumentar en los siguientes ejercicios.

Explique brevemente que es una matriz de *permutación* $P$ y muestre que utilizando una matriz $P$ adecuada se puede resolver este problema y obtener la factorización $PA=LU$. ¿Por qué razón un pivote nulo hace imposible la obtención de la factorización $A=LU$?

```python
import pprint
import numpy as np
import scipy
import scipy.linalg   # SciPy Linear Algebra Library

A = np.array([ [7, 3, -1, 2], [3, 8, 1, -4], [-1, 1, 4, -1], [2, -4, -1, 6] ])
P, L, U = scipy.linalg.lu(A)

pprint.pprint(A) #Matriz A

pprint.pprint(P) #Matriz P

pprint.pprint(L) #Matriz L

pprint.pprint(U) #Matriz U

pprint.pprint(np.dot(L,U))#Matriz A = LU
```
