1. (12 ptos) **Mínimos cuadrados ponderados.**

En mínimos cuadrados, el objetivo (a minimizar) es
$$||A\mathbf{x}-\mathbf{b}||^2=\sum_{i=1}^m (\mathbf{a}_i^T\mathbf{x}-b_i)^2$$
donde los $\mathbf{a}_i$ son las filas de $A$. En el problema de mínimos cuadrados ponderados , minimizamos el objetivo
$$||A\mathbf{x}-\mathbf{b}||^2_W=\sum_{i=1}^m w_i(\mathbf{a}_i^T\mathbf{x}-b_i)^2$$
donde los $w_i$ son pesos positivos. Estos $w_i$  nos permiten asignar diferentes pesos a las diferentes componentes del vector residual.

(a) Demuestre que el objetivo de mínimos cuadrados ponderados se puede expresar como

$$||W(A\mathbf{x}-\mathbf{b})||^2$$

para una matriz diagonal apropiada $W$. Esto nos permite resolver el problema de mínimos cuadrados ponderado  como un problema estándar de mínimos cuadrados, minimizando $$||B\mathbf{x}-\mathbf{d}||^2$$

donde $B = WA$ y $\mathbf{d} = W\mathbf{b}.$

(b) Demuestre o explique por qué cuando $A$ tiene columnas linealmente independientes, también las tiene la matriz $B$.

(c) La solución aproximada de mínimos cuadrados está dada por $𝐱̂=(A^TA)^{-1}A^T\mathbf{b}$. Dé una fórmula similar para la solución del problema de mínimos cuadrados ponderados. Usted podría querer usar la matriz $W = diag(\mathbf{w})$ en su fórmula.

3. (15 ptos) **Método iterativo para el problema de mínimos cuadrados.**

Suponga que las columnas de $A$ forman un conjunto  linealmente independiente, entonces $𝐱̂=(A^TA)^{-1}A^T\mathbf{b}$ minimiza $||A\mathbf{x}-\mathbf{b}||$. En este ejercicio exploramos un método iterativo, debido al matemático *Lewis Richardson*, que se puede utilizar para calcular $𝐱̂$. Definimos
$\mathbf{x}^{(1)} = 0$ y para $k = 1,2,...$

$$\mathbf{x}^{(k+1)}=\mathbf{x}^{(k)}-\mu A^T(A\mathbf{x}^{(k)}-\mathbf{b}),$$

donde $\mu$  es un parámetro positivo y los superíndices indican el número de iteración. Esta define una secuencia de vectores que convergen a $𝐱̂$ siempre que $\mu$  no sea demasiado grande; la elección $\displaystyle\mu=\frac{1}{||A||^2}$, por ejemplo, siempre funciona. La iteración termina cuando $A^T(A\mathbf{x}^{(k)}-\mathbf{b})$
es lo suficientemente pequeño, lo que significa que las condiciones de optimización de mínimos cuadrados están casi satisfechas.
Para implementar el método solo necesitamos multiplicar  vectores por $A$ y por $A^T$. Si tenemos métodos eficientes para llevar a cabo estas dos multiplicaciones matriz-vector, éste método iterativo puede ser más rápido que el algoritmo de solución con la factorización $QR$ (aunque no de la solución exacta).
Los métodos iterativos se utilizan a menudo para problemas de mínimos cuadrados a gran escala.
